```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

# بارگذاری کلید دسترسی از فایل .env
load_dotenv()

# مقداردهی مدل جمینای با دمای صفر (مناسب برای کارهای استدلالی و منطقی)
model = ChatGoogleGenerativeAI(
    model="gemini-flash-lite-latest", 
    temperature=0 
)

print("✅ Gemini model initialized with temperature=0 for reasoning tasks!")
```

```python
# دقت کنید: مثال شامل بخش "استدلال" و "تحلیل" مرحله به مرحله است
examples = """
نام برند: اسنپ
تحلیل:
1. کوتاه و تلفظ آسان است (یک بخشی).
2. معنای سرعت (Snap) را تداعی می‌کند.
3. با ماهیت تاکسی اینترنتی همخوانی دارد.
نتیجه: نام عالی.

نام برند: شرکت حمل و نقل سریع‌السیر برادران و شرکا
تحلیل:
1. بسیار طولانی است (۷ کلمه).
2. به خاطر سپردن آن سخت است.
3. حس قدیمی بودن می‌دهد.
نتیجه: نام ضعیف.
"""

# ساخت قالب پرامپت برای قرار دادن مثال‌ها و نام برند جدید
template = """
تو یک مشاور برندینگ هستی. برای نام برند ورودی، ابتدا مرحله به مرحله تحلیل کن و سپس نتیجه را بگو.
از الگوی زیر پیروی کن:

{examples}

نام برند: {brand_name}
تحلیل:
"""

prompt = ChatPromptTemplate.from_template(template)

print("✅ Examples and Prompt Template created!")
```

```python
# ساخت زنجیره
chain = prompt | model

print("⏳ Analyzing the brand name...\n")

# اجرای مدل
response = chain.invoke({
    "examples": examples,
    "brand_name": "دیجی‌کالا"
})

print("--- 🧠 Chain of Thought Response ---")
print(response.content)
```
