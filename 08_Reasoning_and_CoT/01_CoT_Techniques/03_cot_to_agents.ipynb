```python
# import warnings
# غیرفعال کردن هشدارهای قرمز رنگ LangGraph
# warnings.filterwarnings("ignore")
# 
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

# بارگذاری متغیرهای محیطی
load_dotenv()

# استفاده از مدل gemini-flash-lite-latest که بدون خطا در سیستم شما کار می‌کند
model = ChatGoogleGenerativeAI(
    model="gemini-flash-lite-latest", 
    temperature=0
)

print("✅ Gemini model (flash-lite) initialized for Agent!")
```

```python
from langchain_core.tools import tool

@tool
def get_word_length(word: str) -> int:
    """طول یک کلمه (تعداد حروف آن) را برمی‌گرداند."""
    return len(word)

# لیست ابزارهایی که Agent به آن‌ها دسترسی دارد
tools = [get_word_length]

print("✅ Tool 'get_word_length' created successfully!")
```

```python
from langgraph.prebuilt import create_react_agent

# ۱. ساخت ایجنت با معماری ReAct
agent = create_react_agent(model, tools=tools)

print("⏳ Agent is thinking and acting...\n")

# ۲. ارسال درخواست به ایجنت
response = agent.invoke({
    "messages": [("user", "تعداد حروف کلمه 'LangChain' ضرب در عدد ۳ چند می‌شود؟")]
})

# ۳. چاپ مراحل تفکر و استفاده از ابزار
print("--- 🔍 کالبدشکافی چرخه ReAct (Thought -> Action -> Observation) ---")

for msg in response["messages"]:
    msg_type = type(msg).__name__
    
    if msg_type == "HumanMessage":
        print(f"\n👤 کاربر: {msg.content}")
        
    elif msg_type == "AIMessage" and msg.tool_calls:
        print(f"\n🤖 تفکر مدل (Action): تصمیم به استفاده از ابزار -> {msg.tool_calls[0]['name']}")
        print(f"   ورودی ابزار: {msg.tool_calls[0]['args']}")
        
    elif msg_type == "ToolMessage":
        print(f"\n🛠️ نتیجه ابزار (Observation): {msg.content}")
        
    elif msg_type == "AIMessage" and not msg.tool_calls:
        print(f"\n🤖 تفکر نهایی و جواب (Final Answer): {msg.content}")
```
