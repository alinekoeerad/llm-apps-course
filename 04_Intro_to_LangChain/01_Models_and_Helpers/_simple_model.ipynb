```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, SystemMessage

# Load environment variables securely from the .env file
# This prevents uploading the API key to GitHub
load_dotenv()

# Initialize the Gemini chat model
chat = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.1
)
```

```python
# Define the system role and user prompt
messages = [
    SystemMessage(
        content="تو یک معلم زبان آلمانی هستی. پاسخ دانش‌آموز رو به‌صورت خلاصه و با یک مثال کوتاه رایج و عامیانه بده."
    ),
    HumanMessage(
        content="ترجمه‌ی کلمه‌ی «دانشگاه» به آلمانی چی میشه؟"
    )
]

# Send the request to the model using '.invoke()'
print("⏳ Sending request to the model...")
response = chat.invoke(messages)

# Display the model's text response
print("\n--- Model Response ---")
print(response.content)

# Extract and print token usage metadata
print("\n--- Token Usage Statistics ---")
print(f"Input Tokens:  {response.usage_metadata['input_tokens']}")
print(f"Output Tokens: {response.usage_metadata['output_tokens']}")
print(f"Total Tokens:  {response.usage_metadata['total_tokens']}")
```
