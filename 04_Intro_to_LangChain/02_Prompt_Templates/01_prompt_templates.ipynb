```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

# Load environment variables securely
current_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(current_dir, "..", ".."))
env_path = os.path.join(project_root, ".env")
load_dotenv(dotenv_path=env_path, override=True)

# Initialize the model (Equivalent to ChatOpenAI used in the lesson)
model = ChatGoogleGenerativeAI(
    model="gemini-flash-lite-latest",
    temperature=0.1
)
```

```python
# Define the template using variables inside curly braces {}
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "تو یک دستیار هوشمند هستی که در نقش {role} پاسخ می‌دهی."),
    ("human", "{user_input}")
])

# Inject variables to generate the messages object
messages = prompt_template.invoke({
    "role": "معلم ریاضی شوخ‌طبع",
    "user_input": "عدد ۲ به علاوه ۲ چند می‌شود؟"
})

# Display the structured message objects generated by LangChain
print(messages.messages)
```

```python
# Pass the prepared messages to the model and retrieve the output
response = model.invoke(messages)
print(response.content)
```

```python
# Define a new template specifically for translation
translation_template = ChatPromptTemplate.from_messages([
    ("system", "تو یک مترجم حرفه‌ای هستی. متن کاربر را به زبان {language} ترجمه کن."),
    ("human", "{text}")
])

# Scenario 1: Translate to English
messages_en = translation_template.invoke({
    "language": "انگلیسی",
    "text": "برنامه‌نویسی لذت‌بخش است."
})
print("English:", model.invoke(messages_en).content)
print("-" * 50)

# Scenario 2: Translate to German (Only the variables change, template remains untouched)
messages_de = translation_template.invoke({
    "language": "آلمانی",
    "text": "برنامه‌نویسی لذت‌بخش است."
})
print("German:", model.invoke(messages_de).content)
```
